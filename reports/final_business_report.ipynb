# Customer Spending Behavior & Revenue Analytics

## Final Business Report

### Online Retail II Dataset (Nolan Retail Co.)

---

# 1. Executive Summary

This project analyzes two years of Online Retail II transaction data that we call it Nolan Retail Co. to understand **customer spending behaviour, revenue drivers, cancellations, and customer value**. The analysis combines Python, statistical testing, machine learning, SQL, and Tableau to transform transactional data into actionable business insights.

The results show that revenue is highly concentrated across **customers, geography, and product categories**. Average customer revenue is approximately **£2,916.75**, compared with a median of only **£856.02**, indicating that a relatively small group of high-value customers contributes disproportionately to revenue. The United Kingdom dominates geographic performance, while **Kitchen & Dining** and **Home Decor & Lighting** are the strongest product categories.

Cancellation analysis also reveals meaningful revenue exposure. Some of the company's highest value customers show substantial cancellation activity, and cancellation rates fluctuate over time. Statistical analysis further shows a strong positive association between purchase frequency and customer monetary value.

Customer Lifetime Value modelling provides an additional way to prioritize customers, with the selected KNN regression model achieving an **R² of approximately 0.654**.

Overall, the analysis indicates that the business should focus on **retaining high value repeat customers, reducing financially important cancellations, protecting its strongest revenue drivers, and using customer value analytics to support targeted decision making.**

# 2. Business Objectives

The project aims to answer the following business questions:

* Which customers generate the most revenue?
* How concentrated is revenue across the customer base?
* How does purchase frequency relate to customer value?
* Which countries and product categories generate the most revenue?
* What impact do cancellations have on the business?
* Are high-value customers also associated with significant cancellation activity?
* How does cancellation behaviour change over time?
* Can customer purchasing behaviour help estimate Customer Lifetime Value?

The objective is to translate these findings into recommendations that support **customer retention, revenue protection, product strategy, market prioritization, and customer-value management**.

# 3. Key Findings

## 3.1 Revenue Is Concentrated Among High Value Customers

Customer revenue is strongly right skewed. Average customer revenue is approximately £2,916.75, while the median is only £856.02.

The top 10 customers generate approximately £2.76M, representing 16.14% of customer attributed revenue. Customer 18102 alone generated approximately £580,987 in completed revenue.

This demonstrates that a relatively small group of customers has a substantial influence on overall revenue performance.

## 3.2 Frequent Customers Generate Greater Value

Statistical analysis found a strong positive relationship between purchase frequency and monetary value (Spearman p = 0.859, p < 0.001).

This indicates that customers who purchase more frequently also tend to generate substantially more revenue, making repeat purchasing an important indicator of customer value.

## 3.3 Revenue Is Geographically Concentrated

The United Kingdom generates approximately £14.29M in completed revenue and accounts for approximately 85.5% of revenue.

EIRE, the Netherlands, and Germany are the strongest international markets, although their contributions remain substantially smaller than the UK.

The business therefore has a strong core UK market while also having established international markets that could be evaluated for further growth.

## 3.4 A Few Product Categories Drive Revenue

Kitchen & Dining is the strongest product category at approximately £4.50M, followed by Home Decor & Lighting at £3.97M.

Together, these categories generate approximately £8.47M in completed revenue, showing that product performance is concentrated in a relatively small number of categories.

## 3.5 Cancellations Represent Meaningful Revenue Risk

Cancellation analysis shows that cancellation frequency alone does not capture the full financial impact of cancellations.

Several high value customers also show substantial cancellation activity. For example, Customer 14156 generated approximately £303K in completed revenue with a 24.21% cancellation rate, while Customer 14911 generated approximately £272K with a 23.09% cancellation rate.

Monthly cancellation rates also fluctuate rather than following a consistent upward or downward trend, indicating the importance of monitoring cancellation behaviour over time.

Statistical analysis found that product category and cancellation status are significantly associated, but the relationship is very weak (Cramer's V = 0.046). Product category alone therefore does not adequately explain cancellation behaviour.

## 3.6 Customer Value Can Be Partially Predicted

Customer Lifetime Value modelling compared KNN, Linear Regression, and Decision Tree regression models.

The selected KNN Regressor achieved an R2 of approximately 0.654, meaning that the model explains approximately 65% of the variation in historical customer value in the test data.

The model can therefore support customer prioritization, but prediction errors remain significant for some extreme high-value customers. CLV predictions should be treated as decision support rather than exact financial forecasts.

# 4. Business Recommendations

## 4.1 Prioritize HighValue Repeat Customers

Develop targeted retention strategies for customers with high monetary value and purchase frequency.

Customer prioritization should combine **revenue, purchase frequency, and recency** rather than relying on revenue alone. Loyalty initiatives, personalized offers, and targeted communication can be focused on the customers with the greatest commercial value.

## 4.2 Investigate High Value Customers With High Cancellation Activity

Customers who combine substantial revenue with unusually high cancellation rates should receive priority for further analysis.

The business should investigate whether their cancellations are associated with specific products, large quantities, ordering patterns, stock availability, fulfillment issues, or other operational factors.

Reducing cancellations among valuable customers could protect significant existing revenue.

## 4.3 Monitor Cancellation Rate and Cancelled Revenue

Cancellation performance should be monitored as a recurring business KPI rather than through a single overall percentage.

Management should track both:

* **Monthly cancellation rate**
* **Cancelled revenue**

Unusual increases should then be investigated by customer, product category, order characteristics, and time period.

## 4.4 Protect Strong Product Categories

Kitchen & Dining and Home Decor & Lighting are major revenue drivers.

The company should maintain appropriate inventory availability and commercial support for these categories, particularly during periods of high demand.

Lower performing categories should also be reviewed to determine whether changes to product assortment, promotion, or investment could improve performance.

## 4.5 Protect the UK Core While Evaluating International Opportunities

With approximately **85.5% of revenue generated in the UK**, protecting the domestic customer base should remain a priority.

At the same time, existing international markets such as **EIRE, the Netherlands, and Germany** should be evaluated for targeted growth opportunities rather than pursuing broad international expansion without supporting evidence.

## 4.6 Use Customer Segmentation and CLV to Support Decisions

The average completed customer revenue of approximately **£2,916.75** can be used as an initial benchmark for identifying above-average customers.

However, customer segmentation should ultimately combine **monetary value, frequency, recency, cancellation behaviour, and predicted CLV**.

The CLV model should be used as a supporting prioritization tool alongside historical customer behaviour rather than as an exact financial forecast.

# 5. Limitations

* The analysis is based on historical transactional data, and customer behaviour may change over time.
* Not every transaction contains an identifiable customer, so customer-level analysis applies only to transactions linked to customer IDs.
* The dataset identifies cancellations but does not provide the reason for cancellation. Therefore, the analysis can identify patterns but cannot determine the underlying cause.
* The dataset is heavily concentrated in the UK, so international-market findings should be interpreted with appropriate caution.
* Statistical significance does not necessarily imply strong business importance. For example, product category is significantly associated with cancellation status, but the effect size is very weak.
* CLV predictions are less accurate for extreme high-value customers, so model results should be used for prioritization rather than exact forecasting.


# 6. Conclusion

The analysis shows that revenue performance is driven disproportionately by **high value repeat customers, the UK market, and several strong product categories**. Purchase frequency is strongly associated with customer monetary value, reinforcing the importance of customer retention and repeat purchasing.

Cancellations also represent an important business issue, particularly when they involve high-value customers. Monitoring both cancellation rates and their financial impact can help the business identify where intervention is most valuable.

The strongest business opportunity is therefore to combine **customer segmentation, retention strategies, cancellation monitoring, product performance analysis, and CLV modelling** to prioritize resources toward the customers and activities with the greatest potential impact on revenue.

Overall, the project demonstrates how Python, statistical analysis, machine learning, SQL, and Tableau can be combined to transform transactional retail data into actionable business recommendations.
